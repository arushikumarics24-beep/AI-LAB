```python

from itertools import combinations

def cost(state):
    count = 0
    for i, j in combinations(range(4), 2):
        if abs(state[i] - state[j]) == abs(i - j):
            count += 1
    return count


def hill_climbing(state):
    path_cost = 0

    print("Initial Board:", state)
    print("Initial Cost:", cost(state))

    while True:
        neighbours = []

        for i, j in combinations(range(4), 2):
            new_state = state.copy()
            new_state[i], new_state[j] = new_state[j], new_state[i]
            neighbours.append((cost(new_state), i, j, new_state))

        best = min(neighbours, key=lambda x: x[0])

        if best[0] >= cost(state):
            break

        state = best[3]
        path_cost += 1

        print("Iteration", path_cost, ":", state,
              "Cost:", best[0])

    print("\nFinal Board:", state)
    print("Final Cost:", cost(state))
    print("Path Cost:", path_cost)


initial_state = [3, 1, 2, 0]
hill_climbing(initial_state)
```
