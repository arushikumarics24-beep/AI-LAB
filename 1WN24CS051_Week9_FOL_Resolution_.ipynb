```python

# Resolution in First Order Logic

# A literal is represented as:
# (sign, predicate, arguments)
# sign = True  -> Positive literal
# sign = False -> Negative literal

def unify(x, y, subs):
    if x == y:
        return True

    if x.startswith('?'):
        subs[x] = y
        return True

    if y.startswith('?'):
        subs[y] = x
        return True

    return False


def resolve(c1, c2):
    for l1 in c1:
        for l2 in c2:

            # Predicates must be same and signs opposite
            if l1[0] != l2[0] and l1[1] == l2[1]:

                subs = {}

                if len(l1[2]) != len(l2[2]):
                    continue

                valid = True

                for a, b in zip(l1[2], l2[2]):
                    if not unify(a, b, subs):
                        valid = False
                        break

                if valid:
                    result = []

                    for lit in c1:
                        if lit != l1:
                            result.append(lit)

                    for lit in c2:
                        if lit != l2:
                            result.append(lit)

                    # Apply substitution
                    new_clause = []

                    for sign, pred, args in result:
                        args = tuple(subs.get(a, a) for a in args)
                        new_clause.append((sign, pred, args))

                    return new_clause

    return None


# Given premises converted into CNF

clauses = [

    # John likes all food
    [(False, 'Food', ('?x',)),
     (True, 'Likes', ('John', '?x'))],

    # Anything eaten by someone who is not killed is food
    [(False, 'Eats', ('?y', '?x')),
     (False, 'NotKilled', ('?y',)),
     (True, 'Food', ('?x',))],

    # Anil eats peanuts
    [(True, 'Eats', ('Anil', 'Peanuts'))],

    # Anil is alive
    [(True, 'Alive', ('Anil',))],

    # Anyone alive is not killed
    [(False, 'Alive', ('?x',)),
     (True, 'NotKilled', ('?x',))],

    # Anyone not killed is alive
    [(False, 'NotKilled', ('?x',)),
     (True, 'Alive', ('?x',))],

    # Negation of the conclusion
    [(False, 'Likes', ('John', 'Peanuts'))]
]


# Resolution process

print("Initial CNF Clauses:")
for i, clause in enumerate(clauses, 1):
    print("C", i, ":", clause)

print("\nResolution Steps:")

# Resolve the relevant clauses
r1 = resolve(clauses[3], clauses[4])
print("\nC4 + C5 =", r1)
print("Anil is not killed")

r2 = resolve(clauses[1], clauses[2])
print("\nC2 + C3 =", r2)
print("Food rule requires Anil not killed")

# Resolve Eats and NotKilled with the food rule
r3 = resolve(r2, r1)
print("\nFurther resolution =", r3)
print("Peanuts are food")

# Resolve Food with John's likes-food rule
r4 = resolve(clauses[0], r3)
print("\nFurther resolution =", r4)

# Resolve with negated conclusion
r5 = resolve(r4, clauses[6])
print("\nFinal Resolution =", r5)

if r5 == []:
    print("\nEmpty clause obtained!")
    print("Contradiction found.")
    print("PROVED: John likes peanuts.")
else:
    print("\nProof not completed.")
```
