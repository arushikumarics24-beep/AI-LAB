```python

import random
import math

def heuristic(board):
    conflicts = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def simulated_annealing(n):
    while True:
        board = [random.randint(0, n - 1) for _ in range(n)]
        initial_board = board[:]
        path_cost = 0

        temperature = 100
        cooling_rate = 0.95

        while temperature > 0.01:
            current_h = heuristic(board)

            if current_h == 0:
                return initial_board, board, path_cost

            new_board = board[:]
            col = random.randint(0, n - 1)
            row = random.randint(0, n - 1)

            while row == board[col]:
                row = random.randint(0, n - 1)

            new_board[col] = row

            new_h = heuristic(new_board)
            delta = new_h - current_h

            if delta < 0 or random.random() < math.exp(-delta / temperature):
                board = new_board
                path_cost += 1

                if heuristic(board) == 0:
                    return initial_board, board, path_cost

            temperature *= cooling_rate


def print_board(board):
    n = len(board)

    for i in range(n):
        for j in range(n):
            if board[j] == i:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


n = 8
initial, solution, path_cost = simulated_annealing(n)

print("Initial State:", initial)
print("Initial Cost:", heuristic(initial))
print("\nInitial Board:")
print_board(initial)

print("\nFinal State:", solution)
print("Final Cost:", heuristic(solution))
print("Path Cost:", path_cost)

print("\nFinal Board:")
print_board(solution)
```
