Implement an 8-puzzle problem using
i. DFS
ii. Iterative deepening search (IDS) algorithm


1 DFS

```python
def dfs(state, goal, visited, path, cost):

    if state == goal:
        return path, cost

    visited.add(tuple(state))

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_zero = new_row * 3 + new_col

            new_state = state.copy()

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            if tuple(new_state) not in visited:

                result = dfs(
                    new_state,
                    goal,
                    visited,
                    path + [new_state],
                    cost + 1
                )

                if result:
                    return result

    return None


initial = [1, 2, 3,
           0, 4, 6,
           7, 5, 8]

goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

visited = set()

result = dfs(initial, goal, visited, [initial], 0)

if result:

    path, cost = result

    print("Solution Path:")

    for state in path:
        print(state[0:3])
        print(state[3:6])
        print(state[6:9])
        print()

    print("Path Cost:", cost)

else:
    print("No solution")

print("name:Arushi Kumari")
print("USN:1WN24CS051")
```

2 IDS

```python
# 8-Puzzle using Iterative Deepening Search (IDS)

GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


# Generate possible moves
def get_neighbors(state):

    zero = state.index(0)

    row = zero // 3
    col = zero % 3

    moves = []

    # Move Up
    if row > 0:
        new_state = list(state)
        new_state[zero], new_state[zero - 3] = \
            new_state[zero - 3], new_state[zero]

        moves.append(("Up", tuple(new_state)))

    # Move Down
    if row < 2:
        new_state = list(state)
        new_state[zero], new_state[zero + 3] = \
            new_state[zero + 3], new_state[zero]

        moves.append(("Down", tuple(new_state)))

    # Move Left
    if col > 0:
        new_state = list(state)
        new_state[zero], new_state[zero - 1] = \
            new_state[zero - 1], new_state[zero]

        moves.append(("Left", tuple(new_state)))

    # Move Right
    if col < 2:
        new_state = list(state)
        new_state[zero], new_state[zero + 1] = \
            new_state[zero + 1], new_state[zero]

        moves.append(("Right", tuple(new_state)))

    return moves


# Print puzzle
def print_puzzle(state):

    for i in range(0, 9, 3):
        print(state[i:i + 3])

    print()


# Depth Limited Search
def depth_limited_search(state, depth, path, visited):

    # Goal reached
    if state == GOAL:
        return path

    # Depth limit reached
    if depth == 0:
        return None

    # Generate next states
    for move, next_state in get_neighbors(state):

        if next_state not in visited:

            visited.add(next_state)

            result = depth_limited_search(
                next_state,
                depth - 1,
                path + [move],
                visited
            )

            if result is not None:
                return result

            visited.remove(next_state)

    return None


# IDS function
def ids(start, max_depth=30):

    for depth in range(max_depth + 1):

        print("Searching at depth:", depth)

        visited = {start}

        result = depth_limited_search(
            start,
            depth,
            [],
            visited
        )

        # Goal found
        if result is not None:

            # The depth at which goal is found
            # is the path cost
            return result, depth

    return None


# ---------------- MAIN PROGRAM ----------------

start = (
    1, 2, 3,
    0, 4, 6,
    7, 5, 8
)

print("Initial State:")
print_puzzle(start)

result = ids(start)

if result:

    solution, path_cost = result

    print("\nSolution using IDS:")
    print(" -> ".join(solution))

    print("Path Cost:", path_cost)

else:

    print("No solution found.")


print("name:Arushi Kumari")
print("USN:1WN24CS051")
```
