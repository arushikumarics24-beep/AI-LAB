Implement an 8-puzzle problem using
i. DFS
ii. Iterative deepening search (IDS) algorithm


1 DFS

```python
# 8-Puzzle using Depth First Search (DFS)

GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


# Generate possible moves
def get_neighbors(state):
    zero = state.index(0)

    row = zero // 3
    col = zero % 3

    moves = []

    # Move Up
    if row > 0:
        new_state = list(state)
        new_state[zero], new_state[zero - 3] = \
            new_state[zero - 3], new_state[zero]

        moves.append(("Up", tuple(new_state)))

    # Move Down
    if row < 2:
        new_state = list(state)
        new_state[zero], new_state[zero + 3] = \
            new_state[zero + 3], new_state[zero]

        moves.append(("Down", tuple(new_state)))

    # Move Left
    if col > 0:
        new_state = list(state)
        new_state[zero], new_state[zero - 1] = \
            new_state[zero - 1], new_state[zero]

        moves.append(("Left", tuple(new_state)))

    # Move Right
    if col < 2:
        new_state = list(state)
        new_state[zero], new_state[zero + 1] = \
            new_state[zero + 1], new_state[zero]

        moves.append(("Right", tuple(new_state)))

    return moves


# Print puzzle
def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


# DFS function
def dfs(start):

    # Stack contains:
    # state, path, depth
    stack = [(start, [], 0)]

    visited = set()

    while stack:

        state, path, depth = stack.pop()

        if state in visited:
            continue

        visited.add(state)

        # Goal reached
        if state == GOAL:
            return path, depth

        # Generate next states
        for move, next_state in get_neighbors(state):

            if next_state not in visited:

                stack.append(
                    (next_state,
                     path + [move],
                     depth + 1)
                )

    return None


# ---------------- MAIN PROGRAM ----------------

start = (
    1, 2, 3,
    0, 4, 6,
    7, 5, 8
)

print("Initial State:")
print_puzzle(start)

result = dfs(start)

if result:

    solution, path_cost = result

    print("Solution using DFS:")
    print(" -> ".join(solution))

    print("Path Cost:", path_cost)

else:

    print("No solution found.")

print("name:Arushi Kumari")
print("USN:1WN24CS051")
```

```python
# 8-Puzzle using Iterative Deepening Search (IDS)

GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


# Generate possible moves
def get_neighbors(state):

    zero = state.index(0)

    row = zero // 3
    col = zero % 3

    moves = []

    # Move Up
    if row > 0:
        new_state = list(state)
        new_state[zero], new_state[zero - 3] = \
            new_state[zero - 3], new_state[zero]

        moves.append(("Up", tuple(new_state)))

    # Move Down
    if row < 2:
        new_state = list(state)
        new_state[zero], new_state[zero + 3] = \
            new_state[zero + 3], new_state[zero]

        moves.append(("Down", tuple(new_state)))

    # Move Left
    if col > 0:
        new_state = list(state)
        new_state[zero], new_state[zero - 1] = \
            new_state[zero - 1], new_state[zero]

        moves.append(("Left", tuple(new_state)))

    # Move Right
    if col < 2:
        new_state = list(state)
        new_state[zero], new_state[zero + 1] = \
            new_state[zero + 1], new_state[zero]

        moves.append(("Right", tuple(new_state)))

    return moves


# Print puzzle
def print_puzzle(state):

    for i in range(0, 9, 3):
        print(state[i:i + 3])

    print()


# Depth Limited Search
def depth_limited_search(state, depth, path, visited):

    # Goal reached
    if state == GOAL:
        return path

    # Depth limit reached
    if depth == 0:
        return None

    # Generate next states
    for move, next_state in get_neighbors(state):

        if next_state not in visited:

            visited.add(next_state)

            result = depth_limited_search(
                next_state,
                depth - 1,
                path + [move],
                visited
            )

            if result is not None:
                return result

            visited.remove(next_state)

    return None


# IDS function
def ids(start, max_depth=30):

    for depth in range(max_depth + 1):

        print("Searching at depth:", depth)

        visited = {start}

        result = depth_limited_search(
            start,
            depth,
            [],
            visited
        )

        # Goal found
        if result is not None:

            # The depth at which goal is found
            # is the path cost
            return result, depth

    return None


# ---------------- MAIN PROGRAM ----------------

start = (
    1, 2, 3,
    0, 4, 6,
    7, 5, 8
)

print("Initial State:")
print_puzzle(start)

result = ids(start)

if result:

    solution, path_cost = result

    print("\nSolution using IDS:")
    print(" -> ".join(solution))

    print("Path Cost:", path_cost)

else:

    print("No solution found.")


print("name:Arushi Kumari")
print("USN:1WN24CS051")
```
