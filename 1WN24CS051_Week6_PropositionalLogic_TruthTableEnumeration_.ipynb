```python
from itertools import product

def implies(a, b):
    return not a or b

print("P Q R | Q->P P->~Q QVR | KB")
print("-" * 35)

models = []

for P, Q, R in product([True, False], repeat=3):

    s1 = implies(Q, P)
    s2 = implies(P, not Q)
    s3 = Q or R

    KB = s1 and s2 and s3

    print(P, Q, R, "|", s1, s2, s3, "|", KB)

    if KB:
        models.append((P, Q, R))

print("\nModels of KB:")
for model in models:
    print(model)

def entails(query):
    for P, Q, R in models:
        if not query(P, Q, R):
            return False
    return True

print("\nKB entails R:", entails(lambda P, Q, R: R))
print("KB entails R->P:", entails(lambda P, Q, R: implies(R, P)))
print("KB entails Q->R:", entails(lambda P, Q, R: implies(Q, R)))
```
